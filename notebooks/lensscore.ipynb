```python
import torch
DEVICES = [0] if torch.cuda.is_available() else None

# Quiet PyTorch Lightning logger
import logging, warnings
logging.getLogger("pytorch_lightning.utilities.rank_zero").setLevel(logging.ERROR)
warnings.filterwarnings("ignore")
```

```python
lens_salsa_path = download_model("davidheineman/lens-salsa")
lens_salsa = LENS_SALSA(lens_salsa_path)
```

```python
def calculate_lens_salsa(complex, simple, DEVICES=DEVICES):
    scores, word_level_scores = lens_salsa.score(complex,simple,  batch_size=16, devices=DEVICES)

    return scores
```

```python
CAPTION_FOLDERS = 'E:/Research/gen-captions/gen-captions/scored/'

files = os.listdir(CAPTION_FOLDERS)

base_file = 'gemma3-3-epoch-1000-trained.json'
trained_file = 'gemma3-3-epoch-10000-trained.json'

with open(os.path.join(CAPTION_FOLDERS, base_file), 'r') as f:
    base_captions = json.load(f)

with open(os.path.join(CAPTION_FOLDERS, trained_file), 'r') as f:
    trained_captions = json.load(f)
```

```python
base_captions = [item['model_output'] for item in base_captions]
trained_captions = [item['model_output'] for item in trained_captions]
```

```python
print(len(base_captions), len(trained_captions))
```

```python
scores = calculate_lens_salsa(base_captions, trained_captions, DEVICES)

print(scores)
```
