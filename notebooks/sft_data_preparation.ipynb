```python
import json

train_meta_path = "E:/Research/simplify-me-dataset/train/annotations-train-10000.json"
test_meta_path = "E:/Research/simplify-me-dataset/test/annotations-test.json"
```

```python
with open(train_meta_path, 'r', encoding='utf-8') as f:
    train_annotations = json.load(f)

with open(test_meta_path, 'r', encoding='utf-8') as f:
    test_annotations = json.load(f)
```

```python
print(len(train_annotations))
print(len(test_annotations))
```

```python
print(json.dumps(train_annotations[5], ensure_ascii=False, indent=4))
```

```python
def generate_unique_id(benchmark, img_id):
    return f'{benchmark}-{img_id}'
```

```python
train_data = []

for annotation in train_annotations:
    for caption in annotation['captions']:
        train_data.append(
            {
                'id': annotation["unique_id"],
                'benchmark': annotation['benchmark'],
                'input': "<image>",
                "images": [
                    f"/home/pranon/scratch/def-tahmedge/simplify-me-dataset/simplify-me/compressed_images/{annotation["benchmark"]}/{annotation["image"]["file_name"]}"
                ],
                "instruction": "Describe the image in simple terms",
                "output": caption["caption"]
            }
        )
```
