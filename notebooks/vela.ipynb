````python
PROMPT = """
# Caption Quality and Simplification Evaluation Guide
You are tasked with evaluating candidate caption based on two main categories: **caption quality** and **simplification**. Each category contains multiple criteria that should be scored on a scale from 0 to 1, where 0 is the lowest and 1 is the highest. Always provide reasoning for your scores.

## Caption Quality

Evaluate the overall quality of the caption based on the following criteria:

### 1. Descriptiveness (0-1)
**Definition**: How well does the caption capture and describe the key elements, objects, actions, and context present in the content?

**Scoring Guidelines**:
- **0.0-0.3**: Minimal description; misses most important elements
- **0.4-0.6**: Covers some key elements but lacks detail or misses important aspects
- **0.7-0.8**: Comprehensive description of most important elements with good detail
- **0.9-1.0**: Excellent, thorough description capturing all significant elements with rich detail

**Example**: A caption for an image of a sunset over mountains should mention the sunset, mountains, colors, and any other notable features like clouds or foreground elements.

### 2. Relevance (0-1)
**Definition**: How relevant and accurate is the caption to the actual content? Does it stay focused on what's present without adding irrelevant information?

**Scoring Guidelines**:
- **0.0-0.3**: Mostly irrelevant or inaccurate information; describes wrong content
- **0.4-0.6**: Partially relevant but includes some irrelevant details or minor inaccuracies
- **0.7-0.8**: Highly relevant with minimal irrelevant information
- **0.9-1.0**: Perfectly relevant; every detail mentioned is accurate and pertinent

**Example**: A caption shouldn't mention things that aren't present or make assumptions about context that can't be verified from the content.

### 3. Fluency (0-1)
**Definition**: How natural, grammatically correct, and well-structured is the caption? Does it read smoothly?

**Scoring Guidelines**:
- **0.0-0.3**: Poor grammar, awkward phrasing, difficult to understand
- **0.4-0.6**: Understandable but has grammatical errors or awkward constructions
- **0.7-0.8**: Well-written with natural flow and minor or no grammatical issues
- **0.9-1.0**: Excellent writing quality; perfectly natural and grammatically flawless

**Example**: \"A dog running in park\" (low fluency) vs. \"A dog is running happily in the park\" (high fluency).

### 4. Reasoning
Provide a brief explanation (1-3 sentences) justifying your scores for the caption quality criteria. Highlight what the caption does well and where it falls short.

---

## Simplification

Evaluate how simplified and accessible the caption is based on the following criteria:

### 1. Overall Readability Score (0-1)
**Definition**: A holistic measure of how easy the caption is to read and understand for a general audience.

**Scoring Guidelines**:
- **0.0-0.3**: Very difficult to read; requires advanced comprehension skills
- **0.4-0.6**: Moderately readable; some complexity that may challenge average readers
- **0.7-0.8**: Easily readable by most audiences with good clarity
- **0.9-1.0**: Exceptionally clear and accessible; understandable by all reading levels

**Consider**: Sentence length, vocabulary difficulty, overall complexity.

### 2. Lexical Simplicity (0-1)
**Definition**: How simple and common are the words used? Are technical terms, jargon, or complex vocabulary avoided?

**Scoring Guidelines**:
- **0.0-0.3**: Heavy use of complex, technical, or uncommon words
- **0.4-0.6**: Mix of simple and moderately complex vocabulary
- **0.7-0.8**: Mostly simple, everyday words with minimal complexity
- **0.9-1.0**: Exclusively simple, common words accessible to all readers

**Example**: \"Precipitation\" (complex) vs. \"rain\" (simple); \"utilize\" (complex) vs. \"use\" (simple).

### 3. Syntactical Simplicity (0-1)
**Definition**: How simple is the sentence structure? Are sentences short and straightforward, or long and complex with multiple clauses?

**Scoring Guidelines**:
- **0.0-0.3**: Very complex sentences with multiple clauses, passive voice, or convoluted structure
- **0.4-0.6**: Moderately complex sentences; some compound or complex constructions
- **0.7-0.8**: Mostly simple sentences or well-structured compound sentences
- **0.9-1.0**: Exclusively simple, direct sentence structures that are easy to parse

**Example**: \"The dog, which had been running through the park that was filled with autumn leaves, stopped suddenly\" (low) vs. \"The dog ran through the park and stopped suddenly\" (high).

### 4. Discourse (0-1)
**Definition**: How well-organized and coherent is the caption? Does information flow logically from one point to the next?

**Scoring Guidelines**:
- **0.0-0.3**: Disorganized; ideas jump around without clear connections
- **0.4-0.6**: Somewhat organized but transitions could be smoother
- **0.7-0.8**: Well-organized with logical flow and clear connections
- **0.9-1.0**: Excellently structured with seamless flow and perfect coherence

**Consider** logical progression, use of transitions, and overall organization.

### 5. Stylistic Signals (0-1)
**Definition**: How appropriate and accessible is the writing style? Does it use a straightforward, informative style rather than literary or ornate language?

**Scoring Guidelines**:
- **0.0-0.3**: Highly stylized, literary, or ornate; uses metaphors, idioms, or figurative language heavily
- **0.4-0.6**: Some stylistic elements that may complicate comprehension
- **0.7-0.8**: Mostly straightforward and direct with minimal stylistic flourishes
- **0.9-1.0**: Completely direct and informative; no unnecessary stylistic elements

**Example**: \"The golden orb descended behind the majestic peaks\" (low) vs. \"The sun set behind the mountains\" (high).

### 6. Reasoning
Provide a brief explanation (1-3 sentences) justifying your scores for the simplification criteria. Note what makes the caption easy or difficult to understand.

---

## Output Format

Provide your evaluation in the following JSON structure:

```json
{
  \"caption_quality\": {
    \"descriptiveness\": <score>,
    \"relevance\": <score>,
    \"fluency\": <score>,
    \"reasoning\": \"<your explanation>\"
  },
  \"simplification\": {
    \"overall readability score\": <score>,
    \"lexical simplicity\": <score>,
    \"syntactical simplicity\": <score>,
    \"discourse\": <score>,
    \"stylistic signals\": <score>,
    \"reasoning\": \"<your explanation>\"
  }
}
```

---
"""
````
